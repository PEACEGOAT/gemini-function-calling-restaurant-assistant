```python
from google import genai
from google.genai import types
```

```python
from getpass import getpass

api_key = getpass("Masukkan API Key: ")

client = genai.Client(api_key=api_key)\
```

```python
def get_menu(menu_type: str) -> dict:
    """Returns a menu based on the given menu type."""

    if menu_type == "food":
        return {
            "nasi goreng": 15000,
            "mie goreng": 12000,
            "sate ayam": 20000
        }

    elif menu_type == "drink":
        return {
            "es teh": 5000,
            "es jeruk": 7000,
            "kopi": 10000
        }

    else:
        return {"error": "Menu type not found"}
```

```python
system_instruction = """
Kamu adalah pelayan restoran yang ramah dan profesional.

Tugas kamu:
1. Melayani pelanggan dengan sopan.
2. Memahami pesanan pelanggan.
3. Menggunakan function get_menu untuk mendapatkan harga makanan dan minuman.
4. Memastikan harga berdasarkan data dari function, bukan perkiraan sendiri.
5. Menghitung total harga pesanan dengan tepat.
6. Menampilkan rincian pesanan dan total pembayaran dalam Rupiah.
7. Jika menu tidak ditemukan, sampaikan kepada pelanggan dengan sopan.

Gunakan bahasa Indonesia yang natural dan mudah dipahami.
"""
```

```python
config = types.GenerateContentConfig(
    system_instruction=system_instruction,
    tools=[get_menu]
)
```

```python
response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="Saya pesan Sate ayam, Nasi goreng, dan es teh. Berapa totalnya?",
    config=config
)

print(response.text)
```
